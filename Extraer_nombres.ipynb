```python
import os
from PIL import Image

def detectar_primeras_por_modo(carpeta):
    extensiones_validas = ('.jpg', '.jpeg', '.png', '.webp')

    # modos que queremos detectar
    modos_objetivo = ['RGB', 'RGBA', 'L', '1', 'P', 'LA']

    # diccionario donde guardaremos la primera imagen encontrada de cada modo
    primeras_imagenes = {modo: None for modo in modos_objetivo}

    # recorrer archivos de la carpeta
    for archivo in os.listdir(carpeta):
        ruta = os.path.join(carpeta, archivo)

        if os.path.isfile(ruta) and archivo.lower().endswith(extensiones_validas):
            try:
                with Image.open(ruta) as img:
                    modo = img.mode

                    # si el modo es uno de los buscados y aún no se ha guardado
                    if modo in primeras_imagenes and primeras_imagenes[modo] is None:
                        primeras_imagenes[modo] = ruta

                    # si ya encontró todas, termina
                    if all(primeras_imagenes.values()):
                        break

            except Exception as e:
                print(f"No se pudo procesar {archivo}: {e}")

    return primeras_imagenes


```

```python
# ======= USO =======
carpeta = r"C:\Users\Otro\Documents\Trabajo_grado\Imagenes\webp\train_COCO_data2014_jpg"

resultado = detectar_primeras_por_modo(carpeta)

print("\n=== PRIMERA IMAGEN ENCONTRADA POR MODO ===")
for modo, ruta in resultado.items():
    print(f"{modo}: {ruta}")
```
